```python
import os
from pathlib import Path

# List of directory names
directories = [
    "ArthaBanijya", "Desh", "Manoranjan", "Sahitya", "Swasthya",
    "Bichar", "Khelkud", "Prabas", "SuchanaPrabidhi", "Visw"
]


def combine_text_files():
    with open("combined_simple.txt", 'w', encoding='utf-8') as outfile:
        total_files = 0
        
        
        for directory in directories:
            path = os.path.join("nepalinewsdataset",directory)
            if not os.path.exists(path):
                continue
            
            for root, dirs, files in os.walk(path):
                for file in files:
                    if file.endswith('.txt'):
                        file_path = os.path.join(root, file)
                        try:
                            with open(file_path, 'r', encoding='utf-8') as infile:
                                outfile.write(infile.read())
                                outfile.write('\n\n')
                                total_files += 1
                        except:
                            try:
                                with open(file_path, 'r', encoding='latin-1') as infile:
                                    outfile.write(infile.read())
                                    outfile.write('\n\n')
                                    total_files += 1
                            except:
                                print(f"Could not read: {file_path}")
        
        print(f"\n✅ Done! Combined {total_files} files into 'combined_simple.txt'")

if __name__ == "__main__":
    print("Starting to combine text +files...")
    print("-" * 50)
    

    combine_text_files()
    
    print("\n completed")
```
