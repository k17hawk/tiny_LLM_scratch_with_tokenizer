```python
import os
from pathlib import Path

# List of directory names
directories = [
    "ArthaBanijya", "Desh", "Manoranjan", "Sahitya", "Swasthya",
    "Bichar", "Khelkud", "Prabas", "SuchanaPrabidhi", "Visw"
]


def combine_text_files():
    with open("combined_simple.txt", 'w', encoding='utf-8') as outfile:
        total_files = 0
        
        
        for directory in directories:
            path = os.path.join("nepalinewsdataset",directory)
            if not os.path.exists(path):
                continue
            
            for root, dirs, files in os.walk(path):
                for file in files:
                    if file.endswith('.txt'):
                        file_path = os.path.join(root, file)
                        try:
                            with open(file_path, 'r', encoding='utf-8') as infile:
                                outfile.write(infile.read())
                                outfile.write('\n\n')
                                total_files += 1
                        except:
                            try:
                                with open(file_path, 'r', encoding='latin-1') as infile:
                                    outfile.write(infile.read())
                                    outfile.write('\n\n')
                                    total_files += 1
                            except:
                                print(f"Could not read: {file_path}")
        
        print(f"\n✅ Done! Combined {total_files} files into 'combined_simple.txt'")

if __name__ == "__main__":
    print("Starting to combine text +files...")
    print("-" * 50)
    

    combine_text_files()
    
    print("\n completed")
```

```python
import json


json_files = ['nep_train.json', 'nep_valid.json', 'nep_test.json']

# Output file name
output_file = 'nepali_english_words.txt'

# Set to store unique word pairs (to avoid duplicates)
word_pairs = set()

# Process each JSON file
for json_file in json_files:
    try:
        folder_path = os.path.join("nepali-transliteration",json_file)
        with open(folder_path, 'r', encoding='utf-8') as f:
            # Read each line as separate JSON object
            for line in f:
                line = line.strip()
                if line:  # Skip empty lines
                    data = json.loads(line)
                    native_word = data.get('native word', '')
                    english_word = data.get('english word', '')
                    
                    if native_word and english_word:
                        word_pairs.add(f"{native_word}\t{english_word}")
        
        print(f"Processed: {json_file}")
    except FileNotFoundError:
        print(f"Warning: {json_file} not found")
    except json.JSONDecodeError as e:
        print(f"Error parsing {json_file}: {e}")

# Write to output file
with open(output_file, 'w', encoding='utf-8') as f:
    for pair in sorted(word_pairs):  # Sort alphabetically
        f.write(pair + '\n')

print(f"\nTotal word pairs extracted: {len(word_pairs)}")
print(f"Output saved to: {output_file}")
```

```python
import re
import unicodedata

# Regex: allow Nepali + English + useful symbols
ALLOWED_PATTERN = re.compile(
    r'[^'
    r'a-zA-Z'                    # English letters
    r'\u0900-\u097F'             # Devanagari (Nepali)
    r'0-9\u0966-\u096F'          # Digits (Arabic + Devanagari)
    r'\s'                        # Whitespace
    r'.,!?;:()\[\]{}\-\'\"/\\'   # Basic punctuation
    r'@#$%&+=<>|₹'               # Real-world symbols
    r']+'
)

def clean_for_tokenizer(line: str) -> str:
    """
    Clean a single line for tokenizer training
    (language-agnostic, non-destructive).
    """
    # Unicode normalization (critical for Nepali)
    line = unicodedata.normalize("NFC", line)

    # Remove unwanted characters
    line = ALLOWED_PATTERN.sub(" ", line)

    # Normalize whitespace
    line = re.sub(r"\s+", " ", line)

    return line.strip()


def clean_file(
    input_path,
    output_path,
    min_len=3,
    log_every=1_000_000
):
    print(f"📖 Cleaning {input_path} for tokenizer training...\n")

    total_lines = 0
    kept_lines = 0

    with open(input_path, "r", encoding="utf-8", errors="ignore") as infile, \
         open(output_path, "w", encoding="utf-8") as outfile:

        for line in infile:
            total_lines += 1

            cleaned = clean_for_tokenizer(line)

            if len(cleaned) >= min_len:
                outfile.write(cleaned + "\n")
                kept_lines += 1

            if total_lines % log_every == 0:
                print(f"  Processed {total_lines:,} lines...")

    print("\n✓ Cleaning complete!")
    print(f"Total lines read : {total_lines:,}")
    print(f"Lines kept       : {kept_lines:,}")
    print(f"Lines removed    : {total_lines - kept_lines:,}")
    print(f"Output saved to  : {output_path}")


if __name__ == "__main__":
    clean_file(
        input_path="all_texts_combined.txt",
        output_path="all_texts_combined_clean.txt",
        min_len=3
    )
```

```python
import sentencepiece as spm
import os
from pathlib import Path
from tqdm import tqdm

print("="*70)
print("CHUNKED TRAINING - Guaranteed to work with all 27.5M lines")
print("="*70)

# Configuration
CHUNK_SIZE = 5000000  # 5M lines per chunk (safe for 32GB RAM)
INPUT_FILE = 'all_texts_combined_clean.txt'
MODEL_PREFIX = 'ne_spm_full_1'

# ============================================================================
# STEP 1: Split file into chunks
# ============================================================================
print("\n📂 STEP 1: Splitting file into chunks...")
print(f"   Chunk size: {CHUNK_SIZE:,} lines each\n")

chunk_dir = Path("temp_chunks")
chunk_dir.mkdir(exist_ok=True)

chunk_files = []
total_lines = 0

with open(INPUT_FILE, 'r', encoding='utf-8') as f:
    chunk_num = 0
    
    while True:
        lines = []
        
        # Read chunk_size lines
        for _ in tqdm(range(CHUNK_SIZE), desc=f"Reading chunk {chunk_num}", leave=False):
            line = f.readline()
            if not line:
                break
            lines.append(line)
        
        if not lines:
            break
        
        # Write chunk to file
        chunk_file = chunk_dir / f"chunk_{chunk_num:02d}.txt"
        with open(chunk_file, 'w', encoding='utf-8') as cf:
            cf.writelines(lines)
        
        chunk_files.append(str(chunk_file))
        total_lines += len(lines)
        
        print(f"✓ Chunk {chunk_num}: {len(lines):,} lines → {chunk_file.name}")
        chunk_num += 1

print(f"\n✅ Created {len(chunk_files)} chunks")
print(f"   Total lines: {total_lines:,}")

# ============================================================================
# STEP 2: Train on all chunks
# ============================================================================
print("\n🚀 STEP 2: Training on all chunks...")
print("   This will take 100-140 minutes. Please wait...\n")

try:
    spm.SentencePieceTrainer.train(
        # Pass ALL chunks as comma-separated input
        input=','.join(chunk_files),
        
        model_prefix=MODEL_PREFIX,
        vocab_size=54000,
        model_type='bpe',
        
        # Use ALL data from chunks (no sampling)
        input_sentence_size=0,
        shuffle_input_sentence=True,  # Shuffle across chunks
        
        # Memory-efficient settings
        train_extremely_large_corpus=True,
        num_sub_iterations=2,
        
        max_sentence_length=131072,
        character_coverage=0.9995,
        num_threads=6,  # Conservative to avoid memory spikes
        
        split_by_unicode_script=True,
        split_by_whitespace=True,
        byte_fallback=True,
        
        minloglevel=1
    )
    
    print("\n✅ TRAINING SUCCESSFUL!")
    
except Exception as e:
    print(f"\n❌ Training failed: {e}")
    print("   Trying with even more conservative settings...")
    
    # Fallback: Even more conservative
    spm.SentencePieceTrainer.train(
        input=','.join(chunk_files),
        model_prefix=MODEL_PREFIX,
        vocab_size=54000,
        model_type='bpe',
        
        input_sentence_size=10000000,  # Limit to 15M even from chunks
        shuffle_input_sentence=True,
        train_extremely_large_corpus=True,
        num_sub_iterations=4,  # More iterations = less memory
        
        max_sentence_length=131072,
        character_coverage=0.9995,
        num_threads=4,  # Even fewer threads
        
        split_by_unicode_script=True,
        split_by_whitespace=True,
        byte_fallback=True,
        
        minloglevel=1
    )


print("\n🧹 STEP 3: Cleaning up temporary files...")

for chunk_file in chunk_files:
    os.remove(chunk_file)
    print(f"   Deleted {Path(chunk_file).name}")

chunk_dir.rmdir()

print("\n" + "="*70)
print("✅ COMPLETE!")
print("="*70)
print(f"Model saved: {MODEL_PREFIX}.model")
print(f"Vocab saved: {MODEL_PREFIX}.vocab")
print(f"Trained on: {total_lines:,} lines")
print("="*70)
```

```python
import os
import sentencepiece as spm
from tqdm import tqdm
import threading
import time

#  Force Temp Directory to Disk (Critical for 8.6GB files)
os.environ['TMPDIR'] = '/var/tmp'

print("="*70)
print("ULTRA-LOW-MEMORY TRAINING MODE")
print(f"Input: all_texts_shuffled.txt (8.6GB)")
print("="*70)

class TrainingMonitor:
    def __init__(self, model_prefix):
        self.model_prefix = model_prefix
        self.model_file = f"{model_prefix}.model"
        self.running = True
        self.pbar = tqdm(total=100, desc="Training", unit="%", 
                         bar_format='{l_bar}{bar}| {elapsed}')
        
    def monitor(self):
        last_size = 0
        start_time = time.time()
        while self.running:
            time.sleep(5)
            if os.path.exists(self.model_file):
                current_size = os.path.getsize(self.model_file)
                if current_size > last_size:
                    last_size = current_size
                    elapsed = time.time() - start_time
                    self.pbar.set_postfix({"size": f"{current_size/1024:.1f}KB", 
                                          "time": f"{elapsed:.0f}s"})
        
    def start(self):
        self.thread = threading.Thread(target=self.monitor, daemon=True)
        self.thread.start()
        
    def stop(self):
        self.running = False
        self.pbar.n = 100
        self.pbar.refresh()
        self.pbar.close()

monitor = TrainingMonitor('ne_spm_8gb_safe')
monitor.start()

try:
    spm.SentencePieceTrainer.train(
        input='all_texts_shuffled.txt',
        model_prefix='ne_spm_8gb_safe',
        vocab_size=54000,
        model_type='bpe',
        
        #  Use ALL data
        input_sentence_size=0,
        # File is pre-shuffled
        shuffle_input_sentence=False, 
        # This splits the 8.6GB into 10 chunks of ~860MB each.
        num_sub_iterations=10,      
        max_sentence_length=1024,     
        num_threads=1,               
        character_coverage=0.9995,
        split_by_unicode_script=True,
        split_by_whitespace=True,
        byte_fallback=True,
        minloglevel=1 
    )

finally:
    monitor.stop()

print("\n✅ TRAINING COMPLETE!")
print("   Model: ne_spm_8gb_safe.model")
```

```python
# Independent vowels
VOWELS = {
    "अ": "a",  "आ": "aa", "इ": "i",  "ई": "ii",
    "उ": "u",  "ऊ": "uu", "ए": "e",  "ऐ": "ai",
    "ओ": "o",  "औ": "au", "ऋ": "ri"
}

# Consonants (inherent vowel = 'a')
CONSONANTS = {
    "क": "k",  "ख": "kh", "ग": "g",  "घ": "gh", "ङ": "ng",
    "च": "ch", "छ": "chh","ज": "j",  "झ": "jh", "ञ": "ny",
    "ट": "t",  "ठ": "th", "ड": "d",  "ढ": "dh", "ण": "n",
    "त": "t",  "थ": "th", "द": "d",  "ध": "dh", "न": "n",
    "प": "p",  "फ": "ph", "ब": "b",  "भ": "bh", "म": "m",
    "य": "y",  "र": "r",  "ल": "l",  "व": "w",
    "श": "sh", "ष": "sh", "स": "s",  "ह": "h",
    "क्ष": "ksh","त्र": "tr","ज्ञ": "gy"
}

# Vowel signs (matras)
MATRAS = {
    "ा": "aa", "ि": "i",  "ी": "ii",
    "ु": "u",  "ू": "uu",
    "े": "e",  "ै": "ai",
    "ो": "o",  "ौ": "au"
}

# Other signs
SIGNS = {
    "ं": "n",
    "ँ": "n",
    "ः": "h"
}

HALANT = "्"


def romanize_nepali(text: str) -> str:
    result = []
    i = 0
    length = len(text)

    while i < length:
        ch = text[i]

        # Independent vowel
        if ch in VOWELS:
            result.append(VOWELS[ch])
            i += 1
            continue

        # Consonant
        if ch in CONSONANTS:
            base = CONSONANTS[ch]
            next_char = text[i + 1] if i + 1 < length else ""

            # Halant (no inherent vowel)
            if next_char == HALANT:
                result.append(base)
                i += 2
                continue

            # Matra overrides inherent vowel
            if next_char in MATRAS:
                result.append(base + MATRAS[next_char])
                i += 2
                continue

            # Default inherent vowel
            result.append(base + "a")
            i += 1
            continue

        # Matra without consonant (rare)
        if ch in MATRAS:
            result.append(MATRAS[ch])
            i += 1
            continue

        # Signs
        if ch in SIGNS:
            result.append(SIGNS[ch])
            i += 1
            continue

        # Punctuation / numbers / spaces
        result.append(ch)
        i += 1

    return "".join(result)


# ------------------------
# Example
# ------------------------
if __name__ == "__main__":
    samples = [
        "म आज काठमाडौँ जाँदै छु।",
        "नेपाल एक सुन्दर देश हो।",
        "तिमी के गर्दैछौ?"
    ]

    for s in samples:
        print(s)
        print(romanize_nepali(s))
        print()
```

```python
# -*- coding: utf-8 -*-
"""
Rule-based Nepali (Devanagari) → Romanized English
Safe for large files (line-by-line streaming)
"""

from pathlib import Path

# -----------------------------
# Character mappings
# -----------------------------

# Independent vowels
VOWELS = {
    "अ": "a",  "आ": "aa", "इ": "i",  "ई": "ii",
    "उ": "u",  "ऊ": "uu", "ए": "e",  "ऐ": "ai",
    "ओ": "o",  "औ": "au", "ऋ": "ri"
}

# Consonants (with inherent vowel "a")
CONSONANTS = {
    "क": "k",  "ख": "kh", "ग": "g",  "घ": "gh", "ङ": "ng",
    "च": "ch", "छ": "chh","ज": "j",  "झ": "jh", "ञ": "ny",
    "ट": "t",  "ठ": "th", "ड": "d",  "ढ": "dh", "ण": "n",
    "त": "t",  "थ": "th", "द": "d",  "ध": "dh", "न": "n",
    "प": "p",  "फ": "ph", "ब": "b",  "भ": "bh", "म": "m",
    "य": "y",  "र": "r",  "ल": "l",  "व": "w",
    "श": "sh", "ष": "sh", "स": "s",  "ह": "h"
}

# Vowel signs (matras)
MATRAS = {
    "ा": "aa", "ि": "i",  "ी": "ii",
    "ु": "u",  "ू": "uu",
    "े": "e",  "ै": "ai",
    "ो": "o",  "ौ": "au"
}

# Other marks
SIGNS = {
    "ं": "n",
    "ँ": "n",
    "ः": "h"
}

HALANT = "्"

# -----------------------------
# Romanization function
# -----------------------------

def romanize_nepali(text: str) -> str:
    out = []
    i = 0
    n = len(text)

    while i < n:
        ch = text[i]

        # Independent vowel
        if ch in VOWELS:
            out.append(VOWELS[ch])
            i += 1
            continue

        # Consonant
        if ch in CONSONANTS:
            base = CONSONANTS[ch]
            next_char = text[i + 1] if i + 1 < n else ""

            # Halant → no vowel
            if next_char == HALANT:
                out.append(base)
                i += 2
                continue

            # Matra overrides inherent vowel
            if next_char in MATRAS:
                out.append(base + MATRAS[next_char])
                i += 2
                continue

            # Default inherent vowel
            out.append(base + "a")
            i += 1
            continue

        # Standalone matra (rare)
        if ch in MATRAS:
            out.append(MATRAS[ch])
            i += 1
            continue

        # Anusvara / chandrabindu / visarga
        if ch in SIGNS:
            out.append(SIGNS[ch])
            i += 1
            continue

        # Everything else (English, punctuation, numbers)
        out.append(ch)
        i += 1

    return "".join(out)


# -----------------------------
# File processing
# -----------------------------

def romanize_file(
    input_file: str,
    output_file: str,
    encoding: str = "utf-8"
):
    input_path = Path(input_file)
    output_path = Path(output_file)

    with input_path.open("r", encoding=encoding) as fin, \
         output_path.open("w", encoding=encoding) as fout:

        for line in fin:
            fout.write(romanize_nepali(line))


# -----------------------------
# Run
# -----------------------------

if __name__ == "__main__":
    INPUT_FILE = "combined_simple.txt"
    OUTPUT_FILE = "combined_simple_romanized.txt"

    romanize_file(INPUT_FILE, OUTPUT_FILE)

    print("✅ Romanization finished")
    print(f"📄 Output saved to: {OUTPUT_FILE}")
```

```python
import requests
from bs4 import BeautifulSoup
import time
import json
import csv
from urllib.parse import urljoin
import logging

# Set up logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s'
)
logger = logging.getLogger(__name__)


class NRBFaqScraperEnhanced:
    """
    Enhanced scraper for Nepal Rastra Bank FAQ pages with detailed logging
    and structure validation.
    """
    
    def __init__(self, debug=False):
        self.base_url = "https://www.nrb.org.np"
        self.start_url = "https://www.nrb.org.np/category/faqs/"
        self.session = requests.Session()
        self.session.headers.update({
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
        })
        self.faqs = []
        self.debug = debug
        self.visited_urls = set()
        
    def get_page(self, url):
        """Fetch a page and return BeautifulSoup object with error handling"""
        if url in self.visited_urls:
            logger.info(f"Already visited: {url}")
            return None
            
        try:
            logger.info(f"Fetching: {url}")
            response = self.session.get(url, timeout=30)
            response.raise_for_status()
            self.visited_urls.add(url)
            
            # Polite delay
            time.sleep(1)
            
            return BeautifulSoup(response.content, 'html.parser')
            
        except requests.exceptions.RequestException as e:
            logger.error(f"Error fetching {url}: {e}")
            return None
    
    def extract_post_count(self, div_element):
        """
        Extract post/category count from card-text div
        Looks for pattern: <div class="card-text"><span>X posts</span></div>
        """
        card_text = div_element.find('div', class_='card-text')
        
        if not card_text:
            if self.debug:
                logger.debug("No card-text div found")
            return 0
        
        span = card_text.find('span')
        if not span:
            if self.debug:
                logger.debug("No span in card-text")
            return 0
        
        text = span.get_text().strip()
        if self.debug:
            logger.debug(f"Found span text: '{text}'")
        
        # Extract number from "7 categories" or "2 posts" or "0 posts"
        try:
            count = int(text.split()[0])
            return count
        except (ValueError, IndexError):
            logger.warning(f"Could not parse count from: '{text}'")
            return 0
    
    def is_faq_page(self, soup):
        """
        Determine if current page is an FAQ page (has collapsible cards)
        vs a category listing page
        """
        card_bt_body = soup.find('div', class_='card-bt-body')
        if not card_bt_body:
            return False
        
        # FAQ pages have cards with card-header p-1 class
        faq_cards = card_bt_body.find_all('div', class_='card-header')
        return len(faq_cards) > 0
    
    def scrape_faq_cards(self, soup, url, category_path):
        """Extract FAQs from a page with collapsible card structure"""
        card_bt_body = soup.find('div', class_='card-bt-body')
        if not card_bt_body:
            logger.warning(f"No card-bt-body found on FAQ page: {url}")
            return 0
        
        cards = card_bt_body.find_all('div', class_='card', recursive=False)
        logger.info(f"Found {len(cards)} FAQ cards on page")
        
        extracted = 0
        
        for idx, card in enumerate(cards, 1):
            try:
                # Find card-header with p-1 class
                card_header = card.find('div', class_='card-header')
                if not card_header or 'p-1' not in card_header.get('class', []):
                    if self.debug:
                        logger.debug(f"Card {idx}: No card-header p-1 found")
                    continue
                
                # Get question from h2 > button
                h2 = card_header.find('h2')
                if not h2:
                    if self.debug:
                        logger.debug(f"Card {idx}: No h2 found")
                    continue
                
                button = h2.find('button')
                if not button:
                    if self.debug:
                        logger.debug(f"Card {idx}: No button found")
                    continue
                
                question = button.get_text().strip()
                
                # Find collapse div (class starts with 'collapse-')
                collapse_div = card.find('div', class_=lambda x: x and any('collapse-' in cls for cls in x))
                if not collapse_div:
                    logger.warning(f"Card {idx}: No collapse div found for question: {question[:50]}")
                    continue
                
                # Get answer from card-body > p
                card_body = collapse_div.find('div', class_='card-body')
                if not card_body:
                    logger.warning(f"Card {idx}: No card-body found")
                    continue
                
                # Get all text from card-body (might have multiple p tags)
                answer_parts = []
                for p in card_body.find_all('p'):
                    text = p.get_text().strip()
                    if text:
                        answer_parts.append(text)
                
                answer = '\n\n'.join(answer_parts) if answer_parts else ""
                
                if not answer:
                    logger.warning(f"Card {idx}: Empty answer for question: {question[:50]}")
                
                # Store FAQ
                faq = {
                    'category': category_path,
                    'question': question,
                    'answer': answer,
                    'source_url': url
                }
                
                self.faqs.append(faq)
                extracted += 1
                
                logger.info(f"  ✓ [{extracted}] {question[:60]}...")
                
            except Exception as e:
                logger.error(f"Error processing card {idx}: {e}")
                if self.debug:
                    import traceback
                    traceback.print_exc()
        
        return extracted
    
    def scrape_faq_page_with_pagination(self, url, category_path):
        """Scrape FAQ page and follow pagination"""
        page_num = 1
        total_faqs = 0
        
        current_url = url
        
        while current_url:
            logger.info(f"Scraping page {page_num} of '{category_path}'")
            
            soup = self.get_page(current_url)
            if not soup:
                break
            
            # Extract FAQs from current page
            count = self.scrape_faq_cards(soup, current_url, category_path)
            total_faqs += count
            
            # Look for next page in pagination
            next_url = self.find_next_page(soup, current_url)
            
            if next_url and next_url != current_url:
                current_url = next_url
                page_num += 1
            else:
                break
        
        logger.info(f"Total FAQs from '{category_path}': {total_faqs}")
        return total_faqs
    
    def find_next_page(self, soup, current_url):
        """Find next page URL from pagination block"""
        pagination_block = soup.find('div', class_='pagination-block')
        
        if not pagination_block:
            return None
        
        # Find all page links
        page_links = pagination_block.find_all('a', href=True)
        
        # Find current active page
        for i, link in enumerate(page_links):
            if 'active' in link.get('class', []):
                # Check if there's a next page
                if i + 1 < len(page_links):
                    next_link = page_links[i + 1]
                    # Make sure it's not the current page
                    if 'active' not in next_link.get('class', []):
                        href = next_link.get('href')
                        if href:
                            next_url = urljoin(self.base_url, href)
                            logger.info(f"Found next page: {next_url}")
                            return next_url
                break
        
        return None
    
    def scrape_category_listing(self, url, category_path="FAQs"):
        """Scrape a category listing page"""
        soup = self.get_page(url)
        if not soup:
            return
        
        logger.info(f"\nExploring category: {category_path}")
        logger.info(f"URL: {url}")
        
        # Check if this is actually an FAQ page
        if self.is_faq_page(soup):
            logger.info("This is an FAQ page, not a category listing")
            self.scrape_faq_page_with_pagination(url, category_path)
            return
        
        # Find card-bt-body
        card_bt_body = soup.find('div', class_='card-bt-body')
        if not card_bt_body:
            logger.warning(f"No card-bt-body found in {category_path}")
            return
        
        # Find all category cards matching: col-12 col-lg-3 col-sm-6 col-md-4
        category_divs = card_bt_body.find_all('div', class_=lambda x: x and 
                                               'col-12' in x and 
                                               'col-lg-3' in x and
                                               'col-sm-6' in x and
                                               'col-md-4' in x)
        
        logger.info(f"Found {len(category_divs)} category items")
        
        for idx, div in enumerate(category_divs, 1):
            try:
                # Get post count
                post_count = self.extract_post_count(div)
                
                # Find anchor link
                anchor = div.find('a', href=True)
                if not anchor:
                    logger.warning(f"Item {idx}: No anchor found")
                    continue
                
                link_text = anchor.get_text().strip()
                link_url = urljoin(self.base_url, anchor['href'])
                
                logger.info(f"\n  [{idx}] {link_text}")
                logger.info(f"      Posts/Categories: {post_count}")
                logger.info(f"      URL: {link_url}")
                
                # Only follow if post_count > 0
                if post_count > 0:
                    new_category_path = f"{category_path} > {link_text}"
                    
                    # Recursively scrape (might be another category or FAQ page)
                    self.scrape_category_listing(link_url, new_category_path)
                else:
                    logger.info(f"      Skipping (0 posts/categories)")
                    
            except Exception as e:
                logger.error(f"Error processing category item {idx}: {e}")
                if self.debug:
                    import traceback
                    traceback.print_exc()
    
    def scrape(self):
        """Main scraping entry point"""
        logger.info("="*70)
        logger.info("NRB FAQ SCRAPER - Enhanced Version")
        logger.info("="*70)
        logger.info(f"Starting URL: {self.start_url}")
        logger.info(f"Debug mode: {self.debug}")
        logger.info("="*70)
        
        self.scrape_category_listing(self.start_url)
        
        logger.info("\n" + "="*70)
        logger.info(f"SCRAPING COMPLETE")
        logger.info(f"Total FAQs collected: {len(self.faqs)}")
        logger.info(f"Total pages visited: {len(self.visited_urls)}")
        logger.info("="*70)
        
        return self.faqs
    
    def save_to_json(self, filename='nrb_faqs.json'):
        """Save FAQs to JSON file"""
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(self.faqs, f, ensure_ascii=False, indent=2)
        logger.info(f"Saved {len(self.faqs)} FAQs to {filename}")
    
    def save_to_csv(self, filename='nrb_faqs.csv'):
        """Save FAQs to CSV file"""
        if not self.faqs:
            logger.warning("No FAQs to save")
            return
        
        with open(filename, 'w', newline='', encoding='utf-8') as f:
            writer = csv.DictWriter(f, fieldnames=['category', 'question', 'answer', 'source_url'])
            writer.writeheader()
            writer.writerows(self.faqs)
        logger.info(f"Saved {len(self.faqs)} FAQs to {filename}")
    
    def print_summary(self):
        """Print a summary of collected FAQs"""
        if not self.faqs:
            print("\nNo FAQs collected")
            return
        
        print(f"\n{'='*70}")
        print("FAQ SUMMARY")
        print(f"{'='*70}")
        print(f"Total FAQs: {len(self.faqs)}")
        
        # Group by category
        categories = {}
        for faq in self.faqs:
            cat = faq['category']
            categories[cat] = categories.get(cat, 0) + 1
        
        print(f"\nFAQs by category:")
        for cat, count in sorted(categories.items()):
            print(f"  {cat}: {count} FAQs")
        
        # Show sample FAQs
        print(f"\n{'='*70}")
        print("SAMPLE FAQs (first 3):")
        print(f"{'='*70}")
        
        for i, faq in enumerate(self.faqs[:3], 1):
            print(f"\n{i}. Category: {faq['category']}")
            print(f"   Question: {faq['question']}")
            print(f"   Answer: {faq['answer'][:150]}{'...' if len(faq['answer']) > 150 else ''}")
            print(f"   Source: {faq['source_url']}")


def main():
    """Main execution function"""
    import argparse
    
    parser = argparse.ArgumentParser(description='Scrape FAQs from NRB website')
    parser.add_argument('--debug', action='store_true', help='Enable debug logging')
    parser.add_argument('--output-json', default='nrb_faqs.json', help='Output JSON filename')
    parser.add_argument('--output-csv', default='nrb_faqs.csv', help='Output CSV filename')
    
    args = parser.parse_args()
    
    # Initialize and run scraper
    scraper = NRBFaqScraperEnhanced(debug=args.debug)
    
    # Scrape
    faqs = scraper.scrape()
    
    # Save results
    if faqs:
        scraper.save_to_json(args.output_json)
        scraper.save_to_csv(args.output_csv)
        scraper.print_summary()
    else:
        logger.warning("No FAQs were collected. Check the website structure or enable debug mode.")


if __name__ == "__main__":
    main()
```
